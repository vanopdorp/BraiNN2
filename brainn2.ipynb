```python
import subprocess
import os
import shutil

def clone_repo():
    url = "https://github.com/vanopdorp/BraiNN2.git"
    target = "BraiNN2"

    if os.path.exists(target):
        print(f"Directory '{target}' already exists, removing it...")
        shutil.rmtree(target)

    print("Cloning repository...")
    subprocess.run(["git", "clone", url, target], check=True)
    print("Clone completed.")

def run_script(script_name):
    repo_dir = os.path.join(os.getcwd(), "BraiNN2")
    script_path = os.path.join(repo_dir, script_name)

    subprocess.run(["bash", script_path], cwd=repo_dir, check=True)

def main():
    clone_repo()

    print("\nChoose an option:")
    print("1) Run runbench.sh")
    print("2) Run runtests.sh")
    choice = input("Enter 1 or 2: ").strip()

    if choice == "1":
        run_script("runbench.sh")
    elif choice == "2":
        run_script("runtests.sh")
    else:
        print("Invalid choice.")

if __name__ == "__main__":
    main()
```
